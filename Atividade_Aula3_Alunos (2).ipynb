```python
import pandas as pd
```

```python
path = r"/content/Tabela 1.1.1.xls"
```

```python
formacao_bruto = pd.read_excel(path, header=None)
```

```python
formacao_bruto
```

A tabela deverá ser dividida em 3
- homens e mulheres por região com curso superior
- homens e mulheres por região com curso nas áreas de Ciência, Tecnologia, Engenharias e Matemática
- homens e mulheres por região com curso nas áreas de Educação, Serviços pessoais, Saúde e Bem-estar

Exemplo 1

```python
# : = todas as linhas  |  0:4 = colunas 0, 1, 2 e 3 (as 4 primeiras)
formacao_total = formacao_bruto.iloc[3:36, 0:4].copy()
```

```python
formacao_total.columns = [
    "uf_regiao",
    "total",
    "total_homens",
    "total_mulheres"
]
```

```python
formacao_total
```

```python
formacao_total["percentual_mulheres"] = ((formacao_total['total_mulheres'] / formacao_total['total']) * 100).round(2)
```

```python
# colunas 4, 5 e 6 = Total, Homens e Mulheres da área CTEM
formacao_ctem = formacao_bruto.iloc[3:36, [0, 4, 5, 6]].copy()
```

```python
formacao_ctem.columns = [
    "uf_regiao",
    "total_ctem",
    "homens_ctem",
    "mulheres_ctem"
]
```

```python
formacao_ctem
```

```python
formacao_ctem["percentual_mulheres_ctem"] = ((formacao_ctem['mulheres_ctem'] / formacao_ctem['total_ctem']) * 100).round(2)
```

```python
formacao_ctem
```

```python
# colunas 7, 8 e 9 = Total, Homens e Mulheres da área de Educação, Serviços pessoais, Saúde e Bem-estar
formacao_edu = formacao_bruto.iloc[3:36, [0, 7, 8, 9]].copy()
```
